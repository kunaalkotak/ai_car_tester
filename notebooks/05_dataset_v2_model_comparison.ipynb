## How to read the metrics

- **Accuracy:** fraction of all evaluated examples classified correctly.
- **Precision for AI:** among images predicted AI-generated, the fraction whose label is AI-generated.
- **Recall for AI:** among all AI-generated-labelled images, the fraction the model found.
- **F1-score:** harmonic mean of AI precision and AI recall.
- **ROC-AUC:** ranking quality across a range of thresholds; it does not itself select the application decision threshold.
- **Threshold:** cutoff applied to the sigmoid score to choose class 0 or class 1.
