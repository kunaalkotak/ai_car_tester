## Why compare five approaches?

The first model phase compared four conventional machine-learning methods against a custom CNN:

1. Logistic Regression
2. SVM with an RBF kernel
3. Random Forest
4. Linear Discriminant Analysis (LDA)
5. Custom CNN trained from scratch

The aim was to create accessible baselines before moving to transfer learning.

## Classical-model preprocessing (as recorded for the V1 comparison)

- Convert to RGB, giving 32 × 32 × 3 = **3,072 pixel features**.
- Flatten the RGB array into one feature vector.
